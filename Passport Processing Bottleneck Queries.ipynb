```python
%sql

-- Analyze workload, capacity, and backlog for each adjudication center
SELECT
    Date,
    Processing_Center,
    Applications_In_Process,
    Daily_Capacity,
    Capacity_Utilization_Percent,
    Backlog_Days,
    (Queue_Expedited * 1.0 / (Queue_Standard + Queue_Expedited)) * 100 AS Expedited_Queue_Percent
FROM
    offsite_table3.state_department.capacity_management
WHERE
    Date = (SELECT MAX(Date) FROM offsite_table3.state_department.capacity_management) -- for the latest data
ORDER BY
    Capacity_Utilization_Percent DESC;
```

```python
%sql
SELECT
  `Current_Queue`,
  COUNT(`Application_ID`) AS `Total_Applications`
FROM
  `offsite_table3`.`state_department`.`application_processing`
WHERE
  `Current_Queue` IS NOT NULL
GROUP BY
  `Current_Queue`
```

```python
%sql
SELECT
  `Production_Status`,
  COUNT(`Production_Order_ID`) AS `Total_Orders`,
  SUM(
    CASE
      WHEN `Special_Instructions` ILIKE '%Reprint%' THEN 1
      ELSE 0
    END
  ) AS `Rework_Orders`
FROM
  `offsite_table3`.`printing_companies`.`production_orders`
WHERE
  `Production_Status` IS NOT NULL
GROUP BY
  `Production_Status`
```

```python
%sql
SELECT
  `Processing_Stage`,
  round(AVG(`Days_In_Processing`),2) AS `Average_Days`
FROM
  `offsite_table3`.`state_department`.`application_processing`
WHERE
  `Days_In_Processing` IS NOT NULL
GROUP BY
  `Processing_Stage`
ORDER BY
  `Average_Days` DESC
```
